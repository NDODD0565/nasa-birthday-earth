```python
import requests
import random

def get_epic_photos(api_key, date):
  endpoint_url = f"https://api.nasa.gov/EPIC/api/natural/date/{date}"
  params = {"api_key": api_key}

  response = requests.get(endpoint_url, params=params)

  if response.status_code == 200:
    return response.json()
  else:
    print(f"Data not received. Status code: {response.status_code} ")
    return None

def get_user_input():
  date = input("Enter your birthday (YYYY-MM-DD, 2015 or later): ")
  return date

def birthday_earth(api_key, date):
  photos = get_epic_photos(api_key, date)

  if photos is None:
    print("Could not get photos.")
  elif len(photos) == 0:
    print("No photos found for this date.")
  else:
    print(f"Photos on this date: {len(photos)}")

    photo = random.choice(photos)
    image_name = photo["image"]
    photo_date = photo["date"][:10].replace("-", "/")

    link = f"https://epic.gsfc.nasa.gov/archive/natural/{photo_date}/png/{image_name}.png"

    print("Here is a random Earth photo:")
    print(link)

if __name__ == "__main__":
  api_key = "MQkTrTHch8DcnWzTJhrtKdQmqDKGRwDMQNpwTPgW"
  date = get_user_input()
  birthday_earth(api_key, date)
```
